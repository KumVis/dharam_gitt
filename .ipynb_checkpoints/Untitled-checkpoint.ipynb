```python
import os
import streamlit as st
from dotenv import load_dotenv
from faster_whisper import WhisperModel
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from langchain_groq import ChatGroq
from langchain.schema import HumanMessage, SystemMessage
```

```python
def transcribe_video(video_path: str):
    """Transcribe video into text segments using faster-whisper."""
    model = WhisperModel("small", device="cpu", compute_type="int8")

    # model.transcribe returns (segments generator, info object)
    segments, info = model.transcribe(video_path, beam_size=5)

    texts, timestamps = [], []
    for seg in segments:
        texts.append(seg.text.strip())
        timestamps.append((seg.start, seg.end))

    return texts, timestamps
```

```python
video_path = r"C:\Users\k90vi\Downloads\videoplayback.mp4"
texts, timestamps = transcribe_video(video_path)
```

```python
for t, text in zip(timestamps, texts):
    print(f"{t[0]:.2f} - {t[1]:.2f}: {text}")
```

```python
load_dotenv()
groq_api_key = os.getenv("GROQ_API_KEY")
# Groq LLM (Llama3)
llm = ChatGroq(
    groq_api_key=groq_api_key,
    model_name="openai/gpt-oss-20b"
)

# Embedding model
embedder = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Step 1: Break transcript into chunks (e.g., by sentence or every ~3 lines)
raw_transcript = texts  # original big transcript string
transcript_segments = [seg.strip() for seg in raw_transcript if seg.strip()]  # split by lines

# Step 2: Embed transcript chunks
embeddings = embedder.encode(transcript_segments, convert_to_numpy=True).astype("float32")

# Step 3: Create FAISS index
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)
```

```python
# Step 4: Encode user query
user_question = "What is capital of India"

q_embedding = embedder.encode([user_question], convert_to_numpy=True).astype("float32")

# Step 5: Perform similarity search (top 5 segments for better context)
D, I = index.search(q_embedding, k=5)
retrieved_segments = [transcript_segments[i] for i in I[0]]
context = "\n".join(retrieved_segments)

# Step 6: Create messages for LLM
system_message = SystemMessage(
    content="You are an expert assistant helping to answer questions using only the provided transcript excerpts. "
        "Base your answers strictly on the information in the transcript. "
        "If the answer cannot be found or inferred directly from the transcript, respond with: "
        "'The transcript does not contain this information.' "
        "Do not add extra information or use outside knowledge."
)
human_message = HumanMessage(
    content=f"Transcript context:\n{context}\n\nQuestion: {user_question}"
)

# Step 7: Run LLM
response = llm([system_message, human_message])
print(response.content)
```
